```python
# Import required libraries
from sklearn.datasets import load_diabetes
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import pandas as pd
import matplotlib.pyplot as plt

# Load the Diabetes dataset
diabetes = load_diabetes(as_frame=True)
df = diabetes.frame

# Display dataset
print("Dataset:")
print(df.head())

# Select input features
X = df.drop("target", axis=1)

# Display original dimensions
print("\nOriginal Data Shape:", X.shape)

# Display feature names
print("\nOriginal Features:")
print(X.columns.tolist())

# Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Display reduced dimensions
print("\nReduced Data Shape:", X_pca.shape)

# Display explained variance
print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

print("\nTotal Explained Variance:",
      round(pca.explained_variance_ratio_.sum() * 100, 2), "%")

# Display first five transformed samples
print("\nFirst Five PCA Transformed Samples:")
print(X_pca[:5])

# Create DataFrame for PCA results
pca_df = pd.DataFrame(
    X_pca,
    columns=["Principal Component 1", "Principal Component 2"]
)

print("\nPCA Transformed Dataset:")
print(pca_df.head())

# Visualize PCA result
plt.figure(figsize=(8, 5))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1]
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Dimensionality Reduction - Diabetes Dataset")
plt.grid(True)

plt.show()

# Display final information
print("\nPCA Dimensionality Reduction Completed Successfully")
print("Original Number of Features:", X.shape[1])
print("Reduced Number of Features:", X_pca.shape[1])
```
