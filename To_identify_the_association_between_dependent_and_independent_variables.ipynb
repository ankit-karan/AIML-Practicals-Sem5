```python
from sklearn.datasets import load_diabetes
import pandas as pd

# Load the Diabetes dataset
diabetes = load_diabetes(as_frame=True)

# Convert to DataFrame
df = diabetes.frame

# Display first 5 rows
df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
X = df["bmi"]
Y = df["target"]

print("Independent Variable: BMI")
print("Dependent Variable: Disease Progression (Target)")
```

```python
correlation = X.corr(Y)

print("Pearson Correlation Coefficient:", round(correlation, 4))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.scatter(X, Y)

plt.xlabel("BMI")
plt.ylabel("Disease Progression (Target)")
plt.title("Association Between BMI and Disease Progression")

plt.grid(True)
plt.show()
```

```python
import numpy as np

# Calculate trend line
m, b = np.polyfit(X, Y, 1)

plt.figure(figsize=(8, 5))

plt.scatter(X, Y, label="Data")

plt.plot(X, m * X + b, label="Trend Line")

plt.xlabel("BMI")
plt.ylabel("Disease Progression (Target)")
plt.title("BMI vs Disease Progression")

plt.legend()
plt.grid(True)
plt.show()
```

```python
print("Independent Variable: BMI")
print("Dependent Variable: Target")
print("Correlation Coefficient:", round(correlation, 4))

if correlation > 0:
    print("Interpretation: Positive association")
elif correlation < 0:
    print("Interpretation: Negative association")
else:
    print("Interpretation: No linear association")
```
